```python
import SimpleITK as sitk

# Ruta al archivo
path = "/store/LION/as3628/datasets/raw/LUNA25/luna25_images/1.2.840.113654.2.55.326710919564944433840403373462995060311.mha"
save_img_path = '/store/LION/as3628/datasets/preprocessed/LUNA25/luna25_images/1.3.6.1.4.1.14519.5.2.1.7009.9004.864898009506527801762567164649.mha'
# Leer la imagen
img = sitk.ReadImage(save_img_path)

# Obtener la imagen como array NumPy
import numpy as np
arr = sitk.GetArrayFromImage(img)  # shape = (slices, height, width)
print(type(arr), arr.shape, arr.dtype)  # sanity check
```

```python
import matplotlib
matplotlib.use("Agg")   # must be BEFORE pyplot
import matplotlib.pyplot as plt
slice = arr[20]

plt.figure(figsize=(6,6))
img = plt.imshow(slice, cmap="gray", origin="lower")
plt.colorbar(img)   # <<--- AQUÍ
plt.axis("off")
plt.title("CT slice 100")

plt.savefig("slice100.png", dpi=150, bbox_inches="tight")
plt.close()
```

```python
mask = (slice < -1500) & (slice > -4000)
masked_slice = np.zeros_like(slice)
masked_slice[mask] = slice[mask]

plt.imshow(masked_slice, cmap="gray", origin="lower")
plt.colorbar()
plt.savefig("fondo3.png", dpi=150, bbox_inches="tight")
plt.close()
```
